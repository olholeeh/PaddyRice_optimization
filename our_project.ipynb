```python
import pandas as pd

df = pd.read_csv("paddydataset.csv")
```

```python
print(df.head())
print(df.info())
print(df.describe())
```

```python
df['yield_class'] = pd.qcut(df['Paddy yield(in Kg)'], 3, labels=['Low', 'Medium', 'High'])
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in df.select_dtypes(include=['object', 'string']):
    df[col] = le.fit_transform(df[col])
```

```python
X = df.drop(['Paddy yield(in Kg)', 'yield_class'], axis=1)
y = df['yield_class']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score
import matplotlib.pyplot as plt

depths = [2, 4, 6, 8, 10, 12, None]
scores = []

for d in depths:
    dt = DecisionTreeClassifier(max_depth=d, random_state=42)
    score = cross_val_score(dt, X, y, cv=5, scoring='accuracy').mean()
    scores.append(score)

depth_labels = [str(d) if d is not None else 'None' for d in depths]

plt.plot(depth_labels, scores, marker='o', color='steelblue', linewidth=2)
plt.title('Decision Tree - max_depth vs Accuracy')
plt.xlabel('max_depth')
plt.ylabel('Accuracy')
plt.ylim(0, 1.1)
plt.show()

best_depth = depths[scores.index(max(scores))]
print("Best max_depth:", best_depth)
print("Best Accuracy:", max(scores))
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score
import pandas as pd
import numpy as np

# Start with all features
current_features = list(X.columns)
results = []

while len(current_features) >= 1:
    # Cross validation with current features
    dt = DecisionTreeClassifier(random_state=42)
    score = cross_val_score(dt, X[current_features], y, cv=10, scoring='accuracy').mean()
    results.append({'features': len(current_features), 'accuracy': score})
    print(f"Features: {len(current_features)} — Accuracy: {score:.4f}")
    
    if len(current_features) == 1:
        break
    
    # Compute correlation and keep best V/2 features
    corr = X[current_features].corrwith(pd.Series(y.cat.codes))
    corr_abs = corr.abs().sort_values(ascending=False)
    half = max(1, len(current_features) // 2)
    current_features = list(corr_abs.head(half).index)

# Plot results
feature_counts = [r['features'] for r in results]
accuracies = [r['accuracy'] for r in results]

plt.plot(feature_counts, accuracies, marker='o', color='steelblue', linewidth=2)
plt.title('Feature Selection - Number of Features vs Accuracy')
plt.xlabel('Number of Features')
plt.ylabel('Accuracy')
plt.show()

# Best result
best = max(results, key=lambda x: x['accuracy'])
print("Best number of features:", best['features'])
print("Best accuracy:", best['accuracy'])
```

```python
from sklearn.ensemble import RandomForestClassifier

current_features = list(X.columns)
results_rf = []

while len(current_features) >= 1:
    rf = RandomForestClassifier(random_state=0)
    score = cross_val_score(rf, X[current_features], y, cv=10, scoring='accuracy').mean()
    results_rf.append({'features': len(current_features), 'accuracy': score})
    print(f"Features: {len(current_features)} — Accuracy: {score:.4f}")
    
    if len(current_features) == 1:
        break
    
    corr = X[current_features].corrwith(pd.Series(y.cat.codes))
    corr_abs = corr.abs().sort_values(ascending=False)
    half = max(1, len(current_features) // 2)
    current_features = list(corr_abs.head(half).index)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Get best 11 features from correlation
corr = X.corrwith(pd.Series(y.cat.codes))
best_11 = list(corr.abs().sort_values(ascending=False).head(11).index)
print("Best 11 features:", best_11)

# Split
X_best = X[best_11]
X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(X_best, y, test_size=0.2, random_state=42)

# Decision Tree
dt_best = DecisionTreeClassifier(random_state=42)
dt_best.fit(X_train_b, y_train_b)
y_pred_b = dt_best.predict(X_test_b)
print("Decision Tree Accuracy:", accuracy_score(y_test_b, y_pred_b))
print(classification_report(y_test_b, y_pred_b))

# Random Forest
rf_best = RandomForestClassifier(random_state=0)
rf_best.fit(X_train_b, y_train_b)
rf_pred_b = rf_best.predict(X_test_b)
print("Random Forest Accuracy:", accuracy_score(y_test_b, rf_pred_b))
print(classification_report(y_test_b, rf_pred_b))
```
